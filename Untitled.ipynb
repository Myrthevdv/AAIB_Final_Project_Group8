## 1. Business understanding

The Twente Canal plays a critical role in the logistics infrastructure of the Netherlands by connecting inland regions such as Almelo and Hengelo to major ports like Rotterdam. This makes it an essential corridor for the transportation of goods by barge, which is both cost-effective and environmentally friendly compared to road transport.

However, this waterway is vulnerable to prolonged periods of drought, particularly at the Eefde lock. When the water level at this lock drops below 1.40 meters, the lock becomes inoperable, and the entire Twente Canal becomes inaccessible for freight transport. This disruption can have severe operational and financial consequences for businesses that rely on this corridor.

The goal of this project is to develop a short-term prediction model that forecasts whether water levels at Eefde will fall below this critical threshold in the coming 3 to 10 days. An accurate early warning system would allow logistics service providers, terminal operators, and shipping companies to take timely action, such as rerouting cargo or rescheduling operations, thus improving the resilience and continuity of supply chains in the region.

By framing the challenge from a business perspective, this phase highlights the importance of reliable forecasting not just as a technical task, but as a solution with real-world economic and societal impact.

## 2. Data understanding

### Setup

importing the required libraries and packages

```python
import numpy as np
import pandas as pd
```

We are using 3 different datasets to support our drought prediction model:

1. **Water Level – Eefde Beneden**  
   This dataset contains water level measurements at the Eefde lock, a critical point in the Twente Canal system. The data is sourced from Rijkswaterstaat and originally collected at 10-minute intervals. These values are resampled into daily averages for analysis. Low water levels at this location can block access to the Twente Corridor, making this a key target variable in the project.

2. **Water Level – IJssel**  
   This dataset also comes from Rijkswaterstaat and includes water level data for the IJssel River. Since Eefde is downstream of the IJssel, this data serves as an upstream indicator and is used as an additional predictor for water level forecasting.

3. **Weather Data – Temperature, Humidity, and Precipitation**  
   Weather data is obtained from three regional meteorological stations: Twenthe Airport, Heino, and Deelen Airport. Each station provides daily values for temperature (TG), relative humidity (UG), and precipitation (RH). These environmental factors influence evaporation and inflow, and are therefore used as explanatory variables in the model. The weather data is aggregated and aligned with the water level records for effective feature integration.

## 3. Data preperation

### 3.1 Temperature Data:

missing values and useful columns filtering

```python
#select useful data from the datasets (date, average temperature, average rainfall, average humidity)
temperatureDataTwenthe = temperatureDataTwenthe[['YYYYMMDD', '   TG', '   RH', '   UG']]
temperatureDataHeino = temperatureDataHeino[['YYYYMMDD', '   TG', '   RH', '   UG']]
temperatureDataDeelen = temperatureDataDeelen[['YYYYMMDD', '   TG', '   RH', '   UG']]

#Print one to test whether it is correct
temperatureDataTwenthe.head(10)
```

```python
#change the data YYYYMMDD to the dateTime object to ensure easy queries and combining of multiple notebooks
#first make a copy of the data, otherwise it will only see it as a slice of the data, making future steps impossible
temperatureDataTwenthe = temperatureDataTwenthe.copy()
temperatureDataHeino = temperatureDataHeino.copy()
temperatureDataDeelen = temperatureDataDeelen.copy()

#for each dataset, change the type to datatime, first remove the .0 behind each datetime since it 
temperatureDataTwenthe['YYYYMMDD'] = temperatureDataTwenthe['YYYYMMDD'].astype(str).str.replace('.0', '', regex=False)
temperatureDataTwenthe['YYYYMMDD'] = pd.to_datetime(temperatureDataTwenthe['YYYYMMDD'], errors='coerce')
temperatureDataHeino['YYYYMMDD'] = temperatureDataHeino['YYYYMMDD'].astype(str).str.replace('.0', '', regex=False)
temperatureDataHeino['YYYYMMDD'] = pd.to_datetime(temperatureDataHeino['YYYYMMDD'], errors='coerce')
temperatureDataDeelen['YYYYMMDD'] = temperatureDataDeelen['YYYYMMDD'].astype(str).str.replace('.0', '', regex=False)
temperatureDataDeelen['YYYYMMDD'] = pd.to_datetime(temperatureDataDeelen['YYYYMMDD'], errors='coerce')
```

```python
#only use dates which are within the last 5 years
start_date = pd.Timestamp('2015-05-15')

temperatureDataTwenthe = temperatureDataTwenthe.loc[temperatureDataTwenthe['YYYYMMDD'] >= start_date].reset_index(drop=True)
temperatureDataHeino = temperatureDataHeino.loc[temperatureDataHeino['YYYYMMDD'] >= start_date].reset_index(drop=True)
temperatureDataDeelen = temperatureDataDeelen.loc[temperatureDataDeelen['YYYYMMDD'] >= start_date].reset_index(drop=True)
temperatureDataTwenthe.head(10)
```

```python
#remove rows with missing values for tepmerature since it is missing the least
temperatureDataTwenthe = temperatureDataTwenthe.dropna(subset=['   TG'])
temperatureDataHeino = temperatureDataHeino.dropna(subset=['   TG'])
temperatureDataDeelen = temperatureDataDeelen.dropna(subset=['   TG'])
```

```python
temperatureDataDeelen.isnull().sum()
```

# combine data from all regions

```python
#rename columns to know which region produced the results
df_tw = temperatureDataTwenthe.rename(columns={'   TG': 'TG_Twenthe', '   RH': 'RH_Twenthe', '   UG': 'UG_Twenthe'})
df_he = temperatureDataHeino.rename(columns={'   TG': 'TG_Heino', '   RH': 'RH_Heino', '   UG': 'UG_Heino'})
df_de = temperatureDataDeelen.rename(columns={'   TG': 'TG_Deelen', '   RH': 'RH_Deelen', '   UG': 'UG_Deelen'})

temperatureData = df_tw.merge(df_he, on='YYYYMMDD', how='outer')
temperatureData = temperatureData.merge(df_de, on='YYYYMMDD', how='outer')

temperatureData.head()
```

As can be seen, there are no missing values in this data, However, there are some serious outliers, likely due to failure of equipment. These need to be dealt with.

## 3.2 Eefde bendeden

```python
#this method should remove all rows which have unrealistically high values
waterLevelDataEeefde = waterLevelDataEeefde[waterLevelDataEeefde['waarde'] < 5000]
```

Now that the data is formatted correctly, we can calculate the mean daily water levels. Since the weather data only contained daily values this data needed to be formatted accordingly. To do this, a mean will be taken of the water levels of each day.

```python
#calculate mean water level of each day
dailywaterLevelDataEeefde = waterLevelDataEeefde.set_index('Timestamp').resample('D').mean().reset_index()
dailywaterLevelDataEeefde.head(10)
print(dailywaterLevelDataEeefde.tail(20))
```

# 3.3 Water level data for ijssel

```python
dailywaterLevelDataEeefdeIJssel = waterLevelDataEeefdeIJssel.set_index('Timestamp').resample('D').mean().reset_index()
dailywaterLevelDataEeefdeIJssel.head(10)
```

# combine all data

With all the data ready, it can finally be combined into one large dataset containing all the daily weather data and water level data of the different regions.

```python
# first make all the data into the same format as the weather temperature data
dailywaterLevelDataEeefde.rename(columns={'Timestamp': 'YYYYMMDD', 'waarde': 'WaterLevelEefde'}, inplace=True)
dailywaterLevelDataEeefdeIJssel.rename(columns={'Timestamp': 'YYYYMMDD', 'waarde': 'WaterLevelIJssel'}, inplace=True)

#then combine the datasets to create one large dataset
final_dataset_intermediate = pd.merge(temperatureData, dailywaterLevelDataEeefde, on='YYYYMMDD', how='inner')
final_dataset = pd.merge(final_dataset_intermediate, dailywaterLevelDataEeefdeIJssel, on='YYYYMMDD', how='inner')
final_dataset.rename(columns={'YYYYMMDD': 'date'}, inplace=True)
final_dataset.head(10)
```

# 4. visualization and further DPT

This section will go into depth on the dataset itself, trying to touch upon feature engineering, correlation, scaling, etc. The goal will be to improve the quality of the data, in turn enhancing the performance of the prediction model. The first topic that will be handled is correlation. This is done using a correlation matrix.

```python
#show the correlation matrix between the different columns
#first, some values are still objects, which is why they need to be converted to numerical values
cols_to_convert = ['TG_Twenthe', 'RH_Twenthe', 'UG_Twenthe', 'TG_Heino', 'RH_Heino', 'UG_Heino', 'TG_Deelen', 'RH_Deelen', 'UG_Deelen']  # replace with your actual column names

for col in cols_to_convert:
    final_dataset[col] = pd.to_numeric(final_dataset[col], errors='coerce')

# now the correlation matrix can be created.
import seaborn as sns
sns.heatmap(final_dataset.corr(), annot=True)
```

as can be seen in the correlation matrix, the TG (average temperature) and UG (average humidity) values have a very high correlation value with each other, this is expected since these areas are very close to each other, there are probably not that many significant differences for this value. In order to prevent unnecessary and overrepresented values, both of these values will be replaced by an average of all three columns.


```python
# calculate the average value of temperature and humidity across the different regions
final_dataset['Avg_Temperature'] = final_dataset[['TG_Twenthe', 'TG_Heino', 'TG_Deelen']].mean(axis=1)
final_dataset['Avg_Humidity'] = final_dataset[['UG_Twenthe', 'UG_Heino', 'UG_Deelen']].mean(axis=1)

# drop the unnecessary columns from the dataset
final_dataset = final_dataset.drop(columns=['TG_Twenthe', 'TG_Heino', 'TG_Deelen'])
final_dataset = final_dataset.drop(columns=['UG_Twenthe', 'UG_Heino', 'UG_Deelen'])

final_dataset.head()
```

Now, missing values across the entire dataset can also be checked:

```python
print(final_dataset.isnull().sum())
```

As seen, there are some missing values in the current dataset, mainly in the waterLevelEefde and RH_Twenthe. These values will be added using interpolation, to save valuable data from getting lost.

```python
final_dataset.set_index('date', inplace=True)

# Interpolate numeric columns only:
final_dataset[['RH_Twenthe', 'WaterLevelEefde']] = final_dataset[['RH_Twenthe', 'WaterLevelEefde']].interpolate(method='time')

# Reset index if needed:
final_dataset.reset_index(inplace=True)

print(final_dataset.isnull().sum())
```
